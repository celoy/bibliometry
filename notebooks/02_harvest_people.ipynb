```python
from bibliometry.env import project_root

ROOT = project_root()
```

```python
from bibliometry import pipeline
from bibliometry.config import Config
from bibliometry.env import offer_download

config = Config.load(ROOT / "config.yaml")
config.validate()
client = pipeline.make_client(config)
pipeline.harvest_people_step(config, client)
```

```python
tables = pipeline.tables_step(config)
{name: len(frame) for name, frame in tables.items()}
```

## What is in the tables

* `works` — one row per publication. **`institution_signed`** is true when at least one author signed with the institution; **`roster_people_signed`** lists the people of your list who did;
* `people` — your list, with OpenAlex profiles and publication counts;
* `authorships` — one row per (publication, author, institution);
* `topics`, `keywords`, `funders`, `awards`, `sdgs`, `references`, `counts_by_year`.

They are saved in `data/tables/` (parquet) and summarised in the Excel file below.

```python
people = tables["people"]
people.sort_values("works_signed_with_institution", ascending=False)[
    ["person_id", "lab", "match_status", "works_in_period", "works_signed_with_institution"]
].head(20)
```

```python
excel = sorted((ROOT / "data" / "exports").glob("*.xlsx"))[-1]
offer_download(excel)
```
